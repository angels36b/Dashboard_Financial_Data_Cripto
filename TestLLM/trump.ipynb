```python
import json
import pandas as pd
```

```python
pd.set_option('display.float_format', lambda x: '%.0f' % x)

def inspect_json(filename):
    print(f"=====Inspection :{filename}====")
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            data = json.load(f)
        keys = list(data.keys())
        print(f"Top-level keys: {keys}")

        #define First_key properly before accessing it
        if keys:
            first_key = keys[0]
            print(f"First key contents: {data[first_key]}")

    except FileNotFoundError:
        print(f"Error reading {filename}: [Errno 2] No such file or directory: '{filename}'")
    except Exception as e:
        print(f"Error reading {filename}: {e}")


#Inspect primary text events and baseline rule heuristic files
inspect_json('trump_posts_all.json')
inspect_json('surviving_rules.json')
inspect_json('big_moves.json')
```

Inspecting Inner Key Structures

```python
import json

def inspect_samples():
    #1 Sample post from trump_posts_all.json
    with open('trump_posts_all.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
        posts = data.get('posts', [])
        print("=== Sample Post Object ===")
        if posts:
            print(json.dumps(posts[0], indent=2, ensure_ascii=False))

    #2 Sample Rule from surviving_rules.json
    with open('surviving_rules.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
        rules = data.get('rules', [])
        print("\n=== sample Surviving Rule Object ===")
        if rules:
            print(json.dumps(rules[0], indent=2, ensure_ascii=False))

    # 3 Sample Move Rule from big_moves.json
    with open('big_moves.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
        up_rules = data.get('up_rules', [])
        print("\n=== Sample Up Rule Object ===")
        if up_rules:
            print(json.dumps(up_rules[0], indent=2, ensure_ascii=False))

inspect_samples()
```
